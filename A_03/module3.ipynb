# NumPy Implementation

Each basic gate stores its inputs and weights in NumPy arrays.
It computes a weighted sum and adds a bias:

result = sum(inputs × weights) + bias

The gate returns 1 when the result is positive and 0 otherwise.

AND, NAND, OR, and NOR use different weights and biases.
XOR combines NAND, OR, and AND because a single linear threshold
cannot represent XOR.

```python
import inspect

print(inspect.getsource(LogicGate))
```

# Demonstrating All Five Gates

Each gate is evaluated with (0, 0), (0, 1), (1, 0), and (1, 1).

- AND returns 1 only when both inputs are 1.
- NAND returns the opposite of AND.
- OR returns 1 when at least one input is 1.
- NOR returns the opposite of OR.
- XOR returns 1 when the inputs are different.

```python
input_pairs = [(0, 0), (0, 1), (1, 0), (1, 1)]

functions = [
    ("and_gate", gate.and_gate),
    ("nand_gate", gate.nand_gate),
    ("or_gate", gate.or_gate),
    ("nor_gate", gate.nor_gate),
    ("xor_gate", gate.xor_gate),
]

for name, function in functions:
    print(f"\n{name}")
    for x1, x2 in input_pairs:
        print(f"({x1}, {x2}) -> {function(x1, x2)}")
```
