# Multilayer Perceptron Implementation

# Network Architecture

- Input layer: 3 neurons
- Hidden layer 1: 5 neurons, sigmoid activation
- Hidden layer 2: 5 neurons, sigmoid activation
- Hidden layer 3: 2 neurons, sigmoid activation
- Output layer: 2 neurons, ReLU activation

This notebook demonstrates forward propagation using the
MultilayerPerceptron class.

```python
import numpy as np
from multilayer_perceptron import MultilayerPerceptron

np.set_printoptions(precision=6, suppress=True)

model = MultilayerPerceptron(seed=42)

print("Model created successfully!")
```

# Weights and Biases

Each weight matrix connects one layer to the next.
Each bias vector contains one value for every neuron in its layer.

```python
inputs = np.array([1.0, 0.5, -1.0])

layers = model.forward(inputs, return_layers=True)

print("Input:", inputs)

for number, values in enumerate(layers, start=1):
    activation = "Sigmoid" if number < 4 else "ReLU"
    print(f"Layer {number} ({activation}): {values}")
```
